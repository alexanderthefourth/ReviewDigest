```python
from tabulate import tabulate

lr_metrics = {
    "Accuracy": 0.8600, "Precision": 0.8600,
    "Recall": 0.8600, "F1-score": 0.8600,
    "Negativna P": 0.85, "Negativna R": 0.86, "Negativna F1": 0.86,
    "Pozitivna P": 0.86, "Pozitivna R": 0.85, "Pozitivna F1": 0.86,
}

lstm_metrics = {
    "Accuracy": 0.8480, "Precision": 0.8447,
    "Recall": 0.8526, "F1-score": 0.8486,
    "Negativna P": 0.8434, "Negativna R": 0.8515, "Negativna F1": 0.8474,
    "Pozitivna P": 0.8461, "Pozitivna R": 0.8537, "Pozitivna F1": 0.8499,
}
```

```python
headers = ["Metrika", "LR + TF-IDF", "LSTM (Model 4)", "Razlika (LR−LSTM)"]
rows = []
for key in ["Accuracy", "Precision", "Recall", "F1-score"]:
    diff = lr_metrics[key] - lstm_metrics[key]
    winner = "▲ LR" if diff > 0 else ("▼ LSTM" if diff < 0 else "=")
    rows.append([key, f"{lr_metrics[key]:.4f}", f"{lstm_metrics[key]:.4f}", f"{diff:+.4f}  {winner}"])

print("TABELA 1 – Opšte metrike")
print(tabulate(rows, headers=headers, tablefmt="rounded_outline"))
```

```python
headers2 = ["Klasa / Metrika", "LR + TF-IDF", "LSTM (Model 4)"]
rows2 = [
    ["Negativna – Precision", f"{lr_metrics['Negativna P']:.4f}", f"{lstm_metrics['Negativna P']:.4f}"],
    ["Negativna – Recall",    f"{lr_metrics['Negativna R']:.4f}", f"{lstm_metrics['Negativna R']:.4f}"],
    ["Negativna – F1",        f"{lr_metrics['Negativna F1']:.4f}", f"{lstm_metrics['Negativna F1']:.4f}"],
    ["───────────────────────", "──────────", "──────────────"],
    ["Pozitivna – Precision", f"{lr_metrics['Pozitivna P']:.4f}", f"{lstm_metrics['Pozitivna P']:.4f}"],
    ["Pozitivna – Recall",    f"{lr_metrics['Pozitivna R']:.4f}", f"{lstm_metrics['Pozitivna R']:.4f}"],
    ["Pozitivna – F1",        f"{lr_metrics['Pozitivna F1']:.4f}", f"{lstm_metrics['Pozitivna F1']:.4f}"],
]

print("TABELA 2 – Metrike po klasama")
print(tabulate(rows2, headers=headers2, tablefmt="rounded_outline"))
```

```python
delta_f1 = lr_metrics["F1-score"] - lstm_metrics["F1-score"]
delta_acc = lr_metrics["Accuracy"] - lstm_metrics["Accuracy"]

print("ZAKLJUČCI")
print("─" * 60)
print(f"• LR + TF-IDF postiže F1 = {lr_metrics['F1-score']:.4f}, LSTM (Model 4) postiže F1 = {lstm_metrics['F1-score']:.4f}.")
print(f"• LR nadmašuje LSTM za {delta_f1:+.4f} F1 i {delta_acc:+.4f} Accuracy.")
print(f"• LR je konkurentna referentna tačka uz manji računarski trošak.")
print(f"• LSTM ima potencijal za poboljšanje (Bidirectional, Attention).")
print(f"• Za produkciju sa ograničenim resursima: preporučuje se LR.")
print(f"• Za dalje istraživanje dubokog učenja: LSTM je solidan polazni model.")
```
