```python
import pandas as pd
import numpy as np
import os
import nltk
from collections import Counter
from sklearn.cluster import KMeans
from sentence_transformers import SentenceTransformer
```

```python
df = pd.read_csv('/content/drive/MyDrive/clean_steam_dataset.csv')
print(f'Ucitano recenzija: {len(df)}')
print(df['app_name'].value_counts())
```

```python
nltk.download('punkt')
nltk.download('punkt_tab')

model = SentenceTransformer('all-MiniLM-L6-v2')

games = ['PAYDAY 2', 'Dota 2', 'Grand Theft Auto V', 'Warframe', 'Arma 3']

min_words = 5

for n_clusters in [10, 12]:
  print(f"\n\nBroj klastera: {n_clusters}")
  print("-" * 50)

  for game in games:
    print(f"\n{game}")
    print(f"Broj recenzija: {len(df[df['app_name'] == game])}")

    game_df = df[df['app_name'] == game]
    all_text = " ".join(game_df['review_text'].tolist())
    sentences = nltk.sent_tokenize(all_text)
    sentences = [s.strip() for s in sentences if s.strip() and len(s.split()) >= min_words]

    embeddings = model.encode(sentences, show_progress_bar=True)
    kmeans = KMeans(n_clusters=n_clusters, random_state=42)
    kmeans.fit(embeddings)

    cluster_sizes = Counter(kmeans.labels_)
    sorted_clusters = sorted(range(n_clusters), key=lambda x: cluster_sizes[x], reverse=True)

    representative = []
    for i in sorted_clusters:
      cluster_mask = (kmeans.labels_ == i)
      cluster_embeddings = embeddings[cluster_mask]

      if len(cluster_embeddings) == 0:
        continue

      center = kmeans.cluster_centers_[i]
      distances = np.linalg.norm(cluster_embeddings - center, axis=1)
      closest_idx = np.where(cluster_mask)[0][np.argmin(distances)]
      representative.append(sentences[closest_idx])

    for i, sentence in enumerate(representative, 1):
      print(f"{i}. {sentence}")
```
